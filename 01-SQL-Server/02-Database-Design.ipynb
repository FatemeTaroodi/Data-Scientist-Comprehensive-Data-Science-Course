# Database Design

## 1. What Is Database Design?

**Database Design** is the process of organizing data into a structured database so that the data can be stored, accessed, updated, and maintained efficiently and consistently.

The goal is not simply to create tables. A good database design should answer questions such as:

* What information do we need to store?
* What are the important entities in the system?
* What attributes describe each entity?
* How are entities related to each other?
* How can each record be uniquely identified?
* Where should each piece of information be stored?
* How can we minimize unnecessary data duplication?
* How can we maintain data integrity?

For example, instead of storing all customer, product, employee, and order information in one large table, we identify the different concepts and organize them into related tables.

A simplified process looks like this:

```text
Business Requirements / Raw Data
              ↓
        Identify Entities
              ↓
        Identify Attributes
              ↓
        Identify Keys
              ↓
     Identify Relationships
              ↓
         Normalization
              ↓
        Logical Database
              ↓
        Physical Database
```

---

# 2. Two Main Levels of Database Design

Database design is commonly divided into two major levels:

```text
Database Design
       │
       ├── Logical Design
       │
       └── Physical Design
```

## 2.1 Logical Design

**Logical Design** describes the structure and meaning of the data without focusing on a specific database management system.

At this stage, we determine:

* Entities
* Attributes
* Primary Keys
* Foreign Keys
* Relationships
* Cardinality
* Normalization
* Business rules
* Entity Relationship Diagrams (ERDs)

For example, suppose we are designing a sales system.

At the logical level, we might decide that the system contains:

```text
Customer
Employee
Product
Order
OrderDetail
ProductCategory
```

And that:

```text
Customer 1 ─── * Order

Employee 1 ─── * Order

Order 1 ─── * OrderDetail

Product 1 ─── * OrderDetail

ProductCategory 1 ─── * Product
```

At this stage, we are describing **what the data model looks like**, not exactly how SQL Server or another DBMS will implement it.

---

# 3. Physical Design

**Physical Design** converts the logical model into an actual implementation in a specific DBMS.

For example, after choosing SQL Server, we need to decide:

* Which data types to use
* How to define Primary Keys
* How to define Foreign Keys
* Which columns should be `NOT NULL`
* Which columns should be `UNIQUE`
* Which constraints should be used
* Whether to use `IDENTITY`
* Which indexes should be created
* How to optimize storage and performance

For example:

```sql
CREATE TABLE Customer
(
    CustomerCode INT PRIMARY KEY,
    CustomerName VARCHAR(100) NOT NULL
);
```

The decision that **Customer is an entity** belongs to the logical design.

The decision to implement `CustomerCode` as `INT PRIMARY KEY` in SQL Server belongs to the physical design.

A useful way to remember the difference is:

```text
Logical Design  →  What do we have?
Physical Design →  How do we implement it?
```

---

# 4. Where Do We Start?

There are two major approaches for starting the database design process:

```text
                 Database Design
                       │
              ┌────────┴────────┐
              │                 │
          Top-Down          Bottom-Up
```

The choice depends mainly on whether we are starting with **business requirements** or **existing data**.

---

# 5. Top-Down Approach

In the **Top-Down Approach**, we start from the business requirements and gradually move toward the database structure.

This approach is especially useful when designing a **new system** and there is little or no existing data.

For example, imagine that a company wants a new sales system.

The business requirements might say:

> Customers can place orders. Employees process orders. Each order can contain multiple products. Products belong to categories.

We start by understanding the business:

```text
Business Requirements
          ↓
Business Concepts
          ↓
Entities
          ↓
Attributes
          ↓
Relationships
          ↓
Logical Design
          ↓
Physical Design
          ↓
Database
```

From the requirements, we might identify:

```text
Customer
Employee
Product
ProductCategory
Order
OrderDetail
```

Then we determine their attributes and relationships.

For example:

```text
Customer 1 ─── * Order

Employee 1 ─── * Order

Order 1 ─── * OrderDetail

Product 1 ─── * OrderDetail
```

### When is Top-Down useful?

Top-Down is especially useful when:

* Designing a completely new system
* Existing data does not yet exist
* Business requirements are available
* We need to understand the business process first
* We want the database to reflect the intended business model

---

# 6. Bottom-Up Approach

In the **Bottom-Up Approach**, we start with data that already exists and work toward a better database structure.

The starting point could be:

* Excel files
* CSV files
* Existing database tables
* Legacy systems
* Reports
* Raw datasets

For example, suppose we receive this table:

| OrderCode | CustomerCode | CustomerName | ProductCode | ProductName | EmployeeCode | EmployeeName | Quantity |
| --------- | ------------ | ------------ | ----------- | ----------- | ------------ | ------------ | -------: |
| 100       | 555          | Ali          | 500         | Python      | 666          | Reza         |        2 |
| 100       | 555          | Ali          | 600         | SQL         | 666          | Reza         |        1 |
| 101       | 556          | Sara         | 500         | Python      | 666          | Reza         |        1 |

Instead of designing the system from business requirements, we examine the existing columns and ask:

> What do these attributes describe?

We might discover:

```text
CustomerCode
CustomerName
       ↓
Customer
```

```text
ProductCode
ProductName
       ↓
Product
```

```text
EmployeeCode
EmployeeName
       ↓
Employee
```

```text
OrderCode
Quantity
       ↓
Order / OrderDetail
```

Then we separate the data into appropriate entities and establish relationships.

The general process is:

```text
Existing Raw Data
       ↓
Identify Attributes
       ↓
Group Related Attributes
       ↓
Identify Entities
       ↓
Identify Keys
       ↓
Identify Relationships
       ↓
Normalize
       ↓
Logical Design
       ↓
Physical Design
```

### When is Bottom-Up useful?

Bottom-Up is especially useful when:

* We already have data
* We are migrating a legacy database
* We are cleaning an Excel or CSV dataset
* We need to normalize an existing table
* We are redesigning an existing database

---

# 7. Top-Down vs. Bottom-Up

|                | Top-Down                             | Bottom-Up                                          |
| -------------- | ------------------------------------ | -------------------------------------------------- |
| Starting point | Business requirements                | Existing data                                      |
| Main direction | Business → Database                  | Data → Database                                    |
| Best for       | New systems                          | Existing systems                                   |
| First question | "What does the business need?"       | "What does this data represent?"                   |
| Common source  | Requirements, processes, rules       | Excel, CSV, legacy tables                          |
| Main challenge | Understanding the business correctly | Discovering the intended structure from messy data |

The two approaches are not competitors. In real projects, they are often **combined**.

For example:

```text
Business Requirements
        ↓
    Top-Down
        ↓
Initial Data Model
        ↑
    Bottom-Up
        ↑
Existing Data
```

The business requirements tell us what the system **should represent**, while existing data tells us what information **already exists**.

---

# 8. What Happens After Choosing an Approach?

Regardless of whether we use Top-Down or Bottom-Up, we eventually need to answer the same questions:

```text
1. What are the Entities?
2. What are their Attributes?
3. What are their Keys?
4. How are the Entities related?
5. What is the Cardinality of each relationship?
6. Does the data need to be normalized?
7. What should the Logical Model look like?
8. How should the Logical Model be implemented physically?
```

So the two approaches eventually converge:

```text
              Top-Down
                  │
                  ↓
           ┌─────────────┐
           │   Entities  │
           │  Attributes │
           │    Keys     │
           │ Relationships
           └──────┬──────┘
                  │
                  ↑
              Bottom-Up
```

A practical database design process can be summarized as:

```text
                 START
```

## Finding Entities: The Five-Category Framework

When working Bottom-Up, grouping raw columns into entities gets much easier with this lens — not a strict rule, but a very useful way to think:

| Category | What It Covers | Examples |
|---|---|---|
| **People** | Human roles in the system | Customer, Employee, Student, Author, Member |
| **Objects / Things** | Physical or tangible items | Product, Book, Car, Machine |
| **Concepts** | Intangible but business-relevant ideas | Category, Contract, Account, Membership, Department |
| **Places** | Physical locations | Branch, Warehouse, Store, Address |
| **Events** | Things that *happen*, usually with a date/time and a record of occurrence | Order, Payment, Shipment, Registration, Borrowing |

**A useful (non-absolute) intuition:** an Event often sits at the intersection of People + Objects + Concepts + Places — e.g., an `Order` connects a Customer (People), a Product (Object), maybe a Category (Concept). But an Event doesn't need all four — a simple `Payment` might just need a Customer and an Account.

**How to recognize an Event in raw data:** look for verbs like *ordered, purchased, paid, shipped, borrowed, registered, booked, delivered* — and columns like `Date`, `Time`, `Status`, `Quantity`, `Amount`. But a date column alone isn't proof — `Customer.BirthDate` is a date, but `Customer` itself is still People, not an Event. What matters is whether the *row itself* represents something that *happened*, not just an attribute that happens to be a date.

# Part A — Worked Example: A Sales System (Bottom-Up)

We start from this raw dataset:

````python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
cur = conn.cursor()

cur.execute('''
CREATE TABLE Flat_Sales (
    OrderCode INTEGER, OrderLineCode INTEGER, ProductCode INTEGER, ProductName TEXT,
    CustomerCode INTEGER, CustomerName TEXT, EmployeeCode INTEGER, EmployeeName TEXT,
    CategoryCode INTEGER, CategoryName TEXT, Quantity INTEGER, Price INTEGER, OrderDate TEXT
)
''')
cur.executemany("INSERT INTO Flat_Sales VALUES (?,?,?,?,?,?,?,?,?,?,?,?,?)", [
    (100, 1, 500, "Python", 555, "Ali",     666, "Behnam", 1, "Programming", 1, 1000, "1405-05-05"),
    (100, 2, 600, "SQL",    555, "Ali",     666, "Behnam", 2, "Database",    2, 2000, "1405-05-05"),
    (100, 3, 700, "Excel",  555, "Ali",     666, "Behnam", 3, "Office",      1, 1500, "1405-05-05"),
    (101, 1, 700, "Excel",  556, "Fatemeh", 666, "Behnam", 3, "Office",      1, 1500, "1405-05-06"),
])
conn.commit()
pd.read_sql("SELECT * FROM Flat_Sales", conn)
````

### Step 1–2: Attributes → Entity Classification

| Category | Entity | Attributes |
|---|---|---|
| People | Customer | CustomerCode, CustomerName |
| People | Employee | EmployeeCode, EmployeeName |
| Objects | Product | ProductCode, ProductName, Price, CategoryCode |
| Concepts | ProductCategory | CategoryCode, CategoryName |
| Events | Order | OrderCode, OrderDate, CustomerCode, EmployeeCode |
| Events | OrderDetail | OrderCode, OrderLineCode, ProductCode, Quantity |

`Order` is the Event here — it's the record of "Ali purchased something on this date," bringing together a Customer, an Employee, and (via OrderDetail) Products.

````python
# Logical Design implemented as Physical Design (SQLite syntax)
cur.execute('''CREATE TABLE Customer (
    CustomerCode INTEGER PRIMARY KEY,
    CustomerName TEXT NOT NULL
)''')
cur.executemany("INSERT INTO Customer VALUES (?, ?)", [(555, "Ali"), (556, "Fatemeh")])

cur.execute('''CREATE TABLE Employee (
    EmployeeCode INTEGER PRIMARY KEY,
    EmployeeName TEXT NOT NULL
)''')
cur.executemany("INSERT INTO Employee VALUES (?, ?)", [(666, "Behnam")])

cur.execute('''CREATE TABLE ProductCategory (
    CategoryCode INTEGER PRIMARY KEY,
    CategoryName TEXT NOT NULL
)''')
cur.executemany("INSERT INTO ProductCategory VALUES (?, ?)", [
    (1, "Programming"), (2, "Database"), (3, "Office")
])

cur.execute('''CREATE TABLE Product (
    ProductCode INTEGER PRIMARY KEY,
    ProductName TEXT NOT NULL,
    Price INTEGER NOT NULL CHECK (Price > 0),
    CategoryCode INTEGER,
    FOREIGN KEY (CategoryCode) REFERENCES ProductCategory(CategoryCode)
)''')
cur.executemany("INSERT INTO Product VALUES (?, ?, ?, ?)", [
    (500, "Python", 1000, 1), (600, "SQL", 2000, 2), (700, "Excel", 1500, 3)
])

cur.execute('''CREATE TABLE "Order" (
    OrderCode INTEGER PRIMARY KEY,
    OrderDate TEXT NOT NULL,
    CustomerCode INTEGER,
    EmployeeCode INTEGER,
    FOREIGN KEY (CustomerCode) REFERENCES Customer(CustomerCode),
    FOREIGN KEY (EmployeeCode) REFERENCES Employee(EmployeeCode)
)''')
cur.executemany('INSERT INTO "Order" VALUES (?, ?, ?, ?)', [
    (100, "1405-05-05", 555, 666), (101, "1405-05-06", 556, 666)
])

cur.execute('''CREATE TABLE OrderDetail (
    OrderCode INTEGER,
    OrderLineCode INTEGER,
    ProductCode INTEGER,
    Quantity INTEGER NOT NULL CHECK (Quantity > 0),
    PRIMARY KEY (OrderCode, OrderLineCode),
    FOREIGN KEY (OrderCode) REFERENCES "Order"(OrderCode),
    FOREIGN KEY (ProductCode) REFERENCES Product(ProductCode)
)''')
cur.executemany("INSERT INTO OrderDetail VALUES (?, ?, ?, ?)", [
    (100, 1, 500, 1), (100, 2, 600, 2), (100, 3, 700, 1), (101, 1, 700, 1)
])
conn.commit()
print("Six normalized tables created, with CHECK, PRIMARY KEY, and FOREIGN KEY constraints")
````

````python
# Reconstruct the original flat view to confirm nothing was lost
pd.read_sql('''
    SELECT o.OrderCode, od.OrderLineCode, p.ProductName, c.CustomerName,
           e.EmployeeName, pc.CategoryName, od.Quantity, p.Price, o.OrderDate
    FROM "Order" o
    JOIN OrderDetail od ON o.OrderCode = od.OrderCode
    JOIN Product p ON od.ProductCode = p.ProductCode
    JOIN Customer c ON o.CustomerCode = c.CustomerCode
    JOIN Employee e ON o.EmployeeCode = e.EmployeeCode
    JOIN ProductCategory pc ON p.CategoryCode = pc.CategoryCode
    ORDER BY o.OrderCode, od.OrderLineCode
''', conn)
````

# Part B — Worked Example: A Library System (Top-Down)

This time, no data exists yet — we're told the business rules directly:

> Library has books. Books have authors. Members borrow books. Books are published by publishers. A member can borrow a book multiple times.

### Step 1–2: Entities and Classification

| Category | Entity |
|---|---|
| People | Member, Author |
| Objects | Book |
| Concepts | Publisher |
| Events | Borrowing |

### Why Is `Borrowing` an Event, Not Just a Link Table?

Neither `Member` nor `Book` alone represents *something that happened*. But **"a member borrows a book"** is exactly that — an occurrence, tied to a date. And critically: **the same member can borrow the same book more than once, on different dates.** That single business rule is what turns `Borrowing` from a simple Many-to-Many junction table into a genuine Event table.

### Step 3: Attributes

```text
Member:     MemberCode, MemberName, Phone
Book:       BookCode, BookName, ISBN, Price, PublisherCode
Author:     AuthorCode, AuthorName
Publisher:  PublisherCode, PublisherName
Borrowing:  BorrowingCode, MemberCode, BookCode, BorrowDate, ReturnDate
```

### Step 4: Relationships

```text
Publisher 1 ─── * Book
Book      * ─── * Author        (needs Book_Author junction table)
Member    1 ─── * Borrowing
Book      1 ─── * Borrowing
```

````python
cur.execute('''CREATE TABLE Publisher (
    PublisherCode INTEGER PRIMARY KEY,
    PublisherName TEXT NOT NULL
)''')
cur.executemany("INSERT INTO Publisher VALUES (?, ?)", [(1, "TechPress")])

cur.execute('''CREATE TABLE Book (
    BookCode INTEGER PRIMARY KEY AUTOINCREMENT,
    BookName TEXT NOT NULL,
    ISBN TEXT NOT NULL,
    Price INTEGER NOT NULL,
    PublisherCode INTEGER,
    FOREIGN KEY (PublisherCode) REFERENCES Publisher(PublisherCode)
)''')
cur.execute('INSERT INTO Book (BookCode, BookName, ISBN, Price, PublisherCode) VALUES (1, "Python Basics", "978-0000000001", 1000, 1)')
cur.execute('INSERT INTO Book (BookCode, BookName, ISBN, Price, PublisherCode) VALUES (2, "SQL in Practice", "978-0000000002", 1500, 1)')

cur.execute('''CREATE TABLE Author (
    AuthorCode INTEGER PRIMARY KEY,
    AuthorName TEXT NOT NULL
)''')
cur.executemany("INSERT INTO Author VALUES (?, ?)", [(500, "Vahid Ghorbani"), (501, "Sara Karimi")])

cur.execute('''CREATE TABLE Book_Author (
    BookCode INTEGER,
    AuthorCode INTEGER,
    PRIMARY KEY (BookCode, AuthorCode),
    FOREIGN KEY (BookCode) REFERENCES Book(BookCode),
    FOREIGN KEY (AuthorCode) REFERENCES Author(AuthorCode)
)''')
cur.executemany("INSERT INTO Book_Author VALUES (?, ?)", [(1, 500), (1, 501), (2, 500)])

cur.execute('''CREATE TABLE Member (
    MemberCode INTEGER PRIMARY KEY,
    MemberName TEXT NOT NULL,
    Phone TEXT
)''')
cur.executemany("INSERT INTO Member VALUES (?, ?, ?)", [(1, "Ali Rahmani", "0912...")])

# The Event table
cur.execute('''CREATE TABLE Borrowing (
    BorrowingCode INTEGER PRIMARY KEY AUTOINCREMENT,
    MemberCode INTEGER,
    BookCode INTEGER,
    BorrowDate TEXT NOT NULL,
    ReturnDate TEXT,
    FOREIGN KEY (MemberCode) REFERENCES Member(MemberCode),
    FOREIGN KEY (BookCode) REFERENCES Book(BookCode)
)''')
conn.commit()
print("Library schema created: Publisher, Book, Author, Book_Author, Member, Borrowing")
````

### Proving Why `Borrowing` Needs to Be an Event Table: History

Ali borrows **the same book, twice, on different dates.** If we tried to model this as a simple "current status" column on `Member` or `Book` instead of its own Event table, the second borrowing would silently overwrite the first — we'd lose the history entirely.

````python
cur.executemany("INSERT INTO Borrowing (MemberCode, BookCode, BorrowDate, ReturnDate) VALUES (?, ?, ?, ?)", [
    (1, 1, "1405-01-01", "1405-01-15"),   # Ali borrows Book 1, returns it
    (1, 2, "1405-02-10", "1405-02-20"),   # Ali borrows Book 2
    (1, 1, "1405-03-15", None),           # Ali borrows Book 1 AGAIN — currently out
])
conn.commit()

pd.read_sql('''
    SELECT b.BorrowingCode, m.MemberName, bk.BookName, b.BorrowDate, b.ReturnDate,
           CASE WHEN b.ReturnDate IS NULL THEN 'Still Borrowed' ELSE 'Returned' END AS Status
    FROM Borrowing b
    JOIN Member m ON b.MemberCode = m.MemberCode
    JOIN Book bk ON b.BookCode = bk.BookCode
    ORDER BY b.BorrowDate
''', conn)
````

Because `Borrowing` is its own Event table with a Surrogate Key (`BorrowingCode`), each occurrence gets its own row — Ali's two separate borrowings of Book 1 are both preserved, with different dates and different statuses. This is exactly the point made in the lecture: **Event tables don't just say "Ali currently has Book A" — they record "Ali borrowed Book A on this date and returned it on that date," as a full history.**

````python
# Proving the Book <-> Author Many-to-Many relationship still holds here too
print("Authors per book:")
display(pd.read_sql('''
    SELECT bk.BookName, COUNT(ba.AuthorCode) AS num_authors
    FROM Book bk LEFT JOIN Book_Author ba ON bk.BookCode = ba.BookCode
    GROUP BY bk.BookCode
''', conn))

print("\nHow many times has each book been borrowed (proving Book 1 --- * Borrowing)?")
display(pd.read_sql('''
    SELECT bk.BookName, COUNT(b.BorrowingCode) AS times_borrowed
    FROM Book bk LEFT JOIN Borrowing b ON bk.BookCode = b.BookCode
    GROUP BY bk.BookCode
''', conn))
````

"Python Basics" has been borrowed **2** times by the same member — proof the system correctly preserves history instead of just tracking current state.

# Part C — The Full Reference Checklist

Whenever you're handed a new database problem — a raw dataset or a business description — work through these questions in order.

### A. Business
1. What does the system actually do?
2. What needs to be stored?
3. What events occur in the system?

### B. Entities
4. What People are involved?
5. What Objects/Things exist?
6. What Concepts matter to the business?
7. What Places are relevant?
8. What Events happen?

### C. Attributes
9. What attributes does each entity have?
10. Can any attribute have more than one value per entity instance? (→ needs its own child table)

### D. Keys
11. What's the Primary Key?
12. Is there a natural Business Key?
13. Would a Surrogate Key be better?
14. Is a Composite Key needed?

### E. Relationships
15. Which entities are related?
16. What's the nature of the relationship?
17. Is it 1:1?
18. Is it 1:N?
19. Is it N:M?

### F. Normalization
20. Are all values atomic? (1NF)
21. Any partial dependency? (2NF)
22. Any transitive dependency? (3NF)
23. Any remaining redundant data?

### G. Physical
24. What data type fits each column?
25. What's the Primary Key constraint?
26. What are the Foreign Key constraints?
27. Where is `NOT NULL` required?
28. Where is `UNIQUE` required?
29. Where is `CHECK` useful?
30. Where would an Index help?

## 🔑 Key Takeaways

- **Database Design = Logical Design (what exists) + Physical Design (how to build it).**
- **Top-Down** starts from business requirements (new systems); **Bottom-Up** starts from existing raw data (migrations, normalization exercises).
- The **five-category framework** — People, Objects, Concepts, Places, Events — is the fastest way to turn a wall of raw columns into a set of candidate entities.
- **Events** are the category that most often gets missed: they represent things that *happen*, they usually connect several other entities, and they exist specifically to preserve **History** — the record of every occurrence, not just the current state.
- Every technique from Sessions 5–8 — normalization forms, Business vs. Surrogate Keys, Composite Keys, junction tables for Many-to-Many — fits into this same repeatable pipeline, applied to any new raw dataset or business description you're handed.
