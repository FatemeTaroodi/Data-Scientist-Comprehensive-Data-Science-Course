# Data Entry in SSMS (For Testing)

Although in real projects we usually insert data through applications or controlled scripts,  
SQL Server Management Studio (SSMS) provides a simple way to enter and edit data for testing and learning purposes.

## Method 1: Edit Top 200 Rows (Visual Grid)

1. In **Object Explorer**, right-click on the table.
2. Select **Edit Top 200 Rows**.
3. A grid similar to Excel will open.

In this grid you can:
- Edit existing values
- Insert new rows (type in the empty row at the bottom)
- Delete rows (select the row and press `Delete`)

### Tip
By default, only the first **200** rows are shown.  
To change this limit:
- Click the **SQL** button in the toolbar
- Modify the number `200` to a higher value (e.g., `1000`)

## Method 2: Using INSERT Statements (Recommended)

Even for testing, writing explicit `INSERT` statements is a better practice:

```sql
INSERT INTO Instructor (InstructorID, FirstName, BirthDate)
VALUES (1, N'Ali', '1990-05-15');

INSERT INTO Instructor (InstructorID, FirstName, BirthDate)
VALUES 
    (1, N'Ali', '1990-05-15'),
    (2, N'Sara', '1988-11-22'),
    (3, N'Reza', '1995-03-08');



## Important Notes

| Method                    | Suitable for          | Recommendation                  |
|---------------------------|-----------------------|---------------------------------|
| Edit Top 200 Rows         | Quick testing         | Only for practice / small data  |
| INSERT statements         | Learning + real work  | Preferred method                |
| Application / Forms       | Production            | Best practice                   |

- The visual editor is convenient for small tests.
- For anything beyond simple testing, always prefer writing `INSERT`, `UPDATE`, and `DELETE` statements.
- Be careful when editing data directly — changes are committed immediately.

# Example

In this section we learn how to insert, update, and delete data in SSMS for testing purposes.

We will use a sample table called `Instructor`.

## 1. Create Sample Table

```sql
CREATE TABLE Instructor
(
    InstructorID   INT             NOT NULL,
    FirstName      NVARCHAR(50)    NOT NULL,
    LastName       NVARCHAR(50)    NOT NULL,
    BirthDate      DATE            NULL,
    Email          NVARCHAR(100)   NULL,
    CONSTRAINT PK_Instructor PRIMARY KEY (InstructorID)
);



## 2. Method 1: Edit Top 200 Rows (Visual)

1. Right-click on the table in Object Explorer
2. Choose **Edit Top 200 Rows**
3. A grid similar to Excel will open

You can:
- Type in the empty row at the bottom to insert new data
- Click on any cell to edit existing values
- Select a row and press `Delete` to remove it

**Tip:** To show more than 200 rows, click the **SQL** button in the toolbar and change the number.

## 3. Method 2: Using SQL Statements (Recommended)
