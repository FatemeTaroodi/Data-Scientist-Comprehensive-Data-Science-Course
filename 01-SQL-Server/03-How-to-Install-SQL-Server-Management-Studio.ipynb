## History of SQL Server

The story of **SQL Server** began in the late 1980s, when Microsoft partnered with **Sybase** to develop a relational database system for **OS/2**. The first version, **SQL Server 1.0**, was released in **1989**.

At first, SQL Server was closely related to Sybase's database technology. However, Microsoft gradually moved toward the **Windows NT** platform. In **1993**, SQL Server 4.2 for Windows NT was released, marking an important step in Microsoft's database strategy.

In **1995**, Microsoft released **SQL Server 6.0**, beginning a period in which Microsoft increasingly developed the product independently. Then came **SQL Server 7.0 in 1999**, which was a major rewrite of the database engine and became an important foundation for the SQL Server we know today.

The next major milestone was **SQL Server 2000**. It brought better performance, scalability, clustering, XML support, and stronger enterprise capabilities. In **2005**, SQL Server 2005 introduced many important technologies, including **SQL Server Management Studio (SSMS)**, Dynamic Management Views, CLR integration, XML data types, and Service Broker.

Microsoft continued to improve SQL Server through **2008, 2012, 2014, 2016, 2017, and 2019**, adding better performance, security, high availability, analytics, and Business Intelligence features.

A major change came with **SQL Server 2022**, which strengthened integration with **Azure**, cloud capabilities, analytics, security, and high availability.

As of **2026**, SQL Server has evolved far beyond its original role as a traditional relational database. It is now a mature data platform that supports **transactional workloads, analytics, Business Intelligence, machine learning integration, high availability, security, and hybrid cloud environments**.

### In one sentence

**SQL Server started as a Microsoft–Sybase database for OS/2 in 1989 and evolved over decades into a powerful enterprise data platform deeply integrated with Microsoft's modern data and cloud ecosystem.**


## SQL Server Versions and Editions

Two important concepts in SQL Server are **Version** and **Edition**. They are related, but they describe different things.

### Version

A **Version** identifies the generation or release of SQL Server and usually indicates when it was released and what major technical changes were introduced.

Examples:

* SQL Server 2016
* SQL Server 2017
* SQL Server 2019
* SQL Server 2022

For example:

> **SQL Server 2022**

Here, **2022 is the Version**.

### Edition

An **Edition** defines the level of features, capabilities, and limitations available in a particular SQL Server version.

Common editions include:

| Edition        | Typical Use                       | Main Idea                                              |
| -------------- | --------------------------------- | ------------------------------------------------------ |
| **Express**    | Learning and small applications   | Free and lightweight, but limited                      |
| **Developer**  | Learning and development          | Almost all Enterprise features, but not for Production |
| **Standard**   | Small and medium-sized businesses | Professional features with some limitations            |
| **Enterprise** | Large organizations               | The most complete set of enterprise features           |

For example:

> **SQL Server 2022 Developer Edition**

Here:

* **2022** → Version
* **Developer** → Edition

### Express Edition

**SQL Server Express** is a free and lightweight edition designed for learning, small applications, and small projects.

It is a good choice for:

* Learning SQL
* Student projects
* Small applications
* Practicing database design

However, it has limitations on resources and features compared with higher editions.

### Developer Edition

**SQL Server Developer** is designed for developers, students, and learning environments.

It provides almost all the features available in **Enterprise Edition**, making it an excellent choice for learning and development.

However:

> **Developer Edition is not licensed for Production use.**

You can use it to build and test an application, but a suitable Production edition is required when the system is deployed for real users.

### Standard Edition

**SQL Server Standard** is designed for many business and Production environments.

It provides significantly more capabilities than Express and supports important features for:

* Database management
* Security
* Backup and recovery
* High availability
* Business Intelligence

However, it does not provide all the advanced capabilities of Enterprise Edition.

### Enterprise Edition

**SQL Server Enterprise** provides the most comprehensive set of SQL Server features.

It is designed for large and mission-critical systems where:

* Large amounts of data are involved
* Many users need access
* High availability is important
* High performance is required
* Advanced enterprise features are needed

### Version vs. Edition

A simple way to remember the difference is:

> **Version = Which generation of SQL Server are you using?**
> **Edition = Which level of features and capabilities do you have?**

For example:

```text
SQL Server 2022
       │
       └── Edition
            ├── Express
            ├── Developer
            ├── Standard
            └── Enterprise
```

Therefore, **SQL Server 2022 Developer Edition** means that **2022 is the Version**, while **Developer is the Edition**.


## SQL Server Updates

After releasing a version of SQL Server, Microsoft continues to maintain and support it by releasing **updates, fixes, and security patches**.

These updates can be found in the official **Microsoft Update Catalog**.

For example, the Microsoft Update Catalog provides updates for SQL Server 2025:

[Microsoft Update Catalog – SQL Server 2025](https://www.catalog.update.microsoft.com/Search.aspx?q=sql+server+2025&utm_source=chatgpt.com)

The updates generally include two important categories:

### 1. Security Updates

**Security Updates** are released to fix security vulnerabilities and protect SQL Server from known security threats.

For example, if Microsoft discovers a vulnerability that could allow an attacker to compromise SQL Server, Microsoft releases a **Security Update** to address it.

### 2. Cumulative Updates (CU)

A **Cumulative Update (CU)** contains a collection of fixes and improvements for SQL Server.

Each CU is cumulative, meaning that a newer CU generally includes fixes from previous CUs as well as newer fixes.

For example:

```text
SQL Server 2025
      │
      ├── CU1
      ├── CU2
      ├── CU3
      ├── CU4
      ├── ...
      └── CU8
```

So, after installing a newer CU, you normally do not need to install every previous CU individually.

### Important Point

A SQL Server version does not simply remain unchanged after its initial release. Microsoft continues to maintain it through updates.

A simplified lifecycle looks like this:

```text
SQL Server Version
       │
       ├── Initial Release (RTM)
       │
       ├── Security Updates
       │
       ├── Cumulative Updates (CU)
       │
       └── More fixes and security improvements
```

Therefore, when working with SQL Server, it is important to distinguish between the **SQL Server Version**, the **Edition**, and the **Updates** applied to that version.


### Choosing a SQL Server Version for Production

When deploying SQL Server for a **customer or a Production environment**, it is not always recommended to immediately use the newest available version.

A common practice is to choose a **well-established and stable version** rather than adopting a newly released version immediately.

For example, if a new SQL Server version has recently been released, an organization may prefer to use the previous major version in Production until the new version has been sufficiently tested in real-world environments.

The reason is simple:

> **Production systems prioritize stability and reliability over being on the newest version.**

A new version may introduce new features, but it can also require compatibility testing with:

* Applications
* Drivers
* Third-party software
* Existing databases
* Backup and recovery systems
* Monitoring and administration tools

Therefore, before upgrading a customer's Production environment, the new version should be **tested in a Development or Test environment first**.

### A Typical Approach

```text
New SQL Server Version
          ↓
Development / Test
          ↓
Compatibility & Performance Testing
          ↓
Staging
          ↓
Production
```

This does **not** mean that Production servers should use old or unsupported versions. The selected version should still be **actively supported and regularly patched with security updates and appropriate cumulative updates**.

So the general principle is:

> **Don't choose a version simply because it is the newest. Choose a supported, stable version that has been properly tested for the customer's environment.**

## 9. Feature Selection

SQL Server is not just one component.

During installation, we can choose which SQL Server features we need.

Some important features include:

### Database Engine Services
The core SQL Server component.

It stores data, processes queries, manages transactions, and provides the main relational database functionality.

Allows data to be copied and synchronized between databases or SQL Server instances.

Allows SQL Server to work with languages such as Python and R for certain machine learning scenarios.

Provides advanced searching capabilities for text stored in databases.

Used for analytical and Business Intelligence workloads.

Used to create and manage reports.

In this course we only use the option one.



### Can We Add Features Later?

Yes. You do **not** need to reinstall SQL Server if you later realize that you need another feature.

SQL Server allows you to add features to an existing installation.

For example, suppose we initially install only:

```text
☑ Database Engine Services
☐ Machine Learning Services
☐ Full-Text Search
☐ Integration Services
```

Later, if we need **Full-Text Search**, we can add it to the existing SQL Server installation.

### How to Add a Feature Later

1. Open **SQL Server Installation Center**.
2. Go to **Installation**.
3. Select **New SQL Server stand-alone installation or add features to an existing installation**.
4. Select the existing SQL Server instance.
5. In **Feature Selection**, select the new feature you need.
6. Continue the installation process.

There is no need to reinstall the entire SQL Server instance.

### Feature vs. Database

It is important to distinguish between a **SQL Server Feature** and a **Database**.

A **Feature** is a capability or service provided by SQL Server, such as:

* Database Engine Services
* Full-Text Search
* Integration Services (SSIS)
* Analysis Services (SSAS)
* Machine Learning Services

A **Database**, on the other hand, is created and managed after SQL Server is installed. We can create databases directly through **SQL Server Management Studio (SSMS)** without running the SQL Server installer again.

The general idea is:

```text
SQL Server Installation
        ↓
   Feature Selection
        ↓
   Database Engine
        ↓
   Create Databases
        ↓
       SSMS
```

If we need another feature later:

```text
SQL Server Installation Center
        ↓
Add features to an existing installation
        ↓
Select the new feature
        ↓
Install
```

> **Key Point:** Feature selection is not a permanent decision. You can add additional SQL Server features to an existing installation when they become necessary.

## 13. Authentication Mode

SQL Server supports two main authentication approaches:

### Windows Authentication

Users authenticate using their Windows accounts.

This is generally preferred when the SQL Server environment is integrated with Windows or Active Directory.

### SQL Server and Windows Authentication

Also called **Mixed Mode**.

It allows both:

- Windows Authentication
- SQL Server Authentication

SQL Server Authentication uses a SQL Server login and password.

For example:

`sa`

For many Windows environments, Windows Authentication is preferred because it provides better integration with Windows security.

## What Is a Service?

A service is a program that runs in the background and performs a specific task.

For example, a service can:

* Start automatically when Windows starts
* Run continuously in the background
* Start or stop when needed
* Restart automatically after a failure
* Provide functionality to other applications

Unlike a normal desktop application, a service usually does not have a graphical user interface that the user interacts with directly.

---

## Windows Services vs. Regular Programs

It is useful to distinguish between a **Windows Service** and a regular **program/application**.

### Windows Service

A Windows Service is designed to run in the background and provide a continuous or system-level function.

Examples include:

* SQL Server Database Engine
* Windows Update
* Print Spooler
* Windows Defender services

A service can often start automatically when Windows starts.

### Regular Program

A regular program is usually started by the user and provides a user interface or performs a specific task.

Examples include:

* Microsoft Word
* Google Chrome
* SQL Server Management Studio (SSMS)
* Calculator

For example, when we open SSMS, we interact with it directly through its graphical interface. SSMS is an application, not the SQL Server Database Engine itself.

## SQL Server as a Windows Service

The **SQL Server Database Engine** normally runs as a Windows Service.

For example, in the Services window, we may see:

```text
SQL Server (MSSQLSERVER)
```

or, if we installed a named instance:

```text
SQL Server (SQL2022)
```

This service represents the SQL Server Database Engine running in Windows.

When the service is running, SQL Server can:

* Accept connections from clients
* Execute SQL queries
* Read and write data
* Manage transactions
* Manage databases
* Process requests from applications

If the SQL Server service is stopped, applications such as SSMS cannot connect to that SQL Server instance.

## SQL Server Services and Programs

A common source of confusion is thinking that **SQL Server and SSMS are the same thing**.

They are not.

Think of the architecture like this:

```text
                Windows
                   │
                   ▼
        SQL Server Database Engine
          (Windows Service)
                   │
          ┌────────┴────────┐
          ▼                 ▼
      Databases         SQL Queries
                            ▲
                            │
                           SSMS
                    (Desktop Application)
```

**SQL Server Database Engine** is the actual database server.
**SSMS** is a management application that allows us to connect to and work with the SQL Server.

For example:

```text
SQL Server Service
        ↓
Runs in the background
        ↓
Manages databases and processes queries
        ↑
        │
      SSMS
        ↓
User interface for managing SQL Server
```

### Why Is This Important?

Understanding services becomes especially important when troubleshooting.

For example, if SSMS shows:

```text
Cannot connect to SQL Server
```

one possible reason is that the SQL Server service is not running.

We can check:

```text
Windows Key + R
        ↓
services.msc
        ↓
SQL Server (MSSQLSERVER)
        ↓
Status: Running / Stopped
```

If the service is stopped, SQL Server is not currently running and cannot normally accept connections.

> **Key Point:** SQL Server Database Engine is a server process that commonly runs as a Windows Service, while SSMS is a separate desktop application used to connect to and manage that server.
